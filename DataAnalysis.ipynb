So it's time for us to clean the dataset, one of the things we can notice immediately is G3 is a column with 649 rows of it.

Because at_risk is built off of G3 through:
"combined_dataframe['at_risk'] = (combined_dataframe['G3'] < 10).astype(int)"
... we need to get rid of G3 so it can never be a model input. It would hand the model an answer and that would be called 'target leakage'.




```python
import pandas as pd
loc = "data/dataProject.csv"

df = pd.read_csv(loc)
print(df)
```
