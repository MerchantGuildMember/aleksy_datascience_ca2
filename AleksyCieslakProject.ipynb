## Import pandas and the fetch from ucimlrepo

```python
from ucimlrepo import fetch_ucirepo
import pandas as pd
```

```python
# Fetch the repo and store the object it returns
fetched_data_object = fetch_ucirepo(id=320)
fetched_data_object.variables
```

```python
# Combine .data.features and .data.targets into a DataFrame
combined_dataframe = pd.concat([fetched_data_object.data.features, fetched_data_object.data.targets], axis=1)
print(combined_dataframe.shape)
combined_dataframe.info()
```

```python
# Note that seems to be the Portugeuse course only
# Add a column for whether a student is determined to be at_risk if their third grade is lower than 10
# Sets at_risk to 1 if deemed to be at risk of failing, 0 otherwise.
combined_dataframe['at_risk'] = (combined_dataframe['G3'] < 10).astype(int)
failing_count = combined_dataframe['at_risk'].value_counts()
print(failing_count)
```
